```python
#1.Question 1:A retail store wants to identify customers who make frequent purchases. 
import warnings
warnings.filterwarnings("ignore")
import pandas as pd

# Dataset
data = {
    'Customer_ID': [101, 102, 103, 101, 104, 102, 101, 105, 102, 103],
    'Purchase_Amount': [200, 150, 180, 220, 300, 200, 100, 400, 250, 300]
}

# Create DataFrame
df = pd.DataFrame(data)

# 1️⃣ Group by Customer_ID and calculate total purchase amount
total_purchase = df.groupby('Customer_ID')['Purchase_Amount'].sum()

# 2️⃣ Sort to find top 3 customers
top_customers = total_purchase.sort_values(ascending=False).head(3)

print("Total purchase per customer:\n", total_purchase)
print("\nTop 3 customers with highest purchase amounts:\n", top_customers)
```

```python
#2. Predicting House Prices with Linear Regression
import pandas as pd
from sklearn.linear_model import LinearRegression

# Dataset
data = {
    'Square_Feet': [1500, 2000, 2500, 3000, 3500],
    'Price': [300000, 400000, 500000, 600000, 700000]
}

# Convert to DataFrame
df = pd.DataFrame(data)

# Split into features (X) and target (y)
X = df[['Square_Feet']]
y = df['Price']

# Train Linear Regression model
model = LinearRegression()
model.fit(X, y)

# Test data
test_data = [[1800], [2800]]

# Predict
predicted_prices = model.predict(test_data)

print("Predicted Prices:")
print(predicted_prices)
```

```python
#3 Set Question
import pandas as pd

data = {'Category': ['A', 'B', 'C', 'A', 'D', 'B', 'A', 'E', 'B', 'C', 'C', 'A'],
        'Value': [10, 15, 20, 30, 25, 18, 22, 40, 35, 50, 45, 15]}

df = pd.DataFrame(data)

# Count occurrences per category
category_counts = df['Category'].value_counts().reset_index()
category_counts.columns = ['Category', 'Count']

# Identify the top 3 most frequent categories
top3 = category_counts.head(3)

print("Total Occurrences per Category:")
print(category_counts)
print("\nTop 3 Frequent Categories:")
print(top3)
```

```python
#4 Set Question

import pandas as pd

data = {'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'Age': [25, 30, None, 35, None]}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

# Calculate the mean of the Age column (excluding NaN)
mean_age = df['Age'].mean()

# Replace missing values (NaN) with the mean
df['Age'].fillna(mean_age, inplace=True)

print("\nData after Imputation:")
print(df)
```

```python
#5: Linear Regression: Predict Salary
import pandas as pd
from sklearn.linear_model import LinearRegression

# Data
data = {'Experience': [1, 2, 3, 4, 5], 'Salary': [30000, 35000, 40000, 45000, 50000]}
df = pd.DataFrame(data)

# Prepare features and target
X = df[['Experience']]
y = df['Salary']

# Train model
model = LinearRegression()
model.fit(X, y)

# Predict for 6 years of experience
experience_new = pd.DataFrame([[6]], columns=['Experience'])
predicted_salary = model.predict(experience_new)[0]

print(f"Predicted Salary for 6 years of experience: ${predicted_salary:.2f}")
```

```python
#6: Simple Classification: Spam/Ham Prediction 
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

# Dataset
data = {'word_count': [100, 150, 200, 120, 180, 220], 
        'is_spam': ['ham', 'spam', 'spam', 'ham', 'spam', 'spam']}
df = pd.DataFrame(data)

# Features and target
X = df[['word_count']]
y = df['is_spam']

# Train model
clf = DecisionTreeClassifier()
clf.fit(X, y)

# Predict for email with 200 words
prediction = clf.predict([[120]])[0]
print(f"Prediction for email with 200 words: {prediction.capitalize()}")
```
