```python
import pandas as pd
import numpy as np

data = {'Feature1': ['10', '20', 'Thirty'],  # 'Thirty' is non-numeric
        'Feature2': [5.5, 6.7, 8.9]}

##ANSWER-1:  Fill Non Numeric with 'NaN'
##----------------------------------------------------
df = pd.DataFrame(data)
df['Feature1'] = pd.to_numeric(df['Feature1'], errors='coerce')
##Remarks:
## In Pandas, the parameter errors='coerce' (used in pd.to_numeric()) means:
## If a value cannot be converted to a number, replace it with NaN (Not a Number) instead of raising an error.
print("With 'NaN' values\n",df,"\n")

##ANSWER-2:  Fill 'Non Numeric' with '0'
##----------------------------------------------------
df['Feature1'].fillna(0, inplace=True)
print("Replacing NaN with '0':\n",df)
```

```python
#5.
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder

X = [[1, 2], [3, 4], [5, 6]]
y = ['yes', 'no', 'yes']  # Error: Labels should be numeric

##Approach-1
y1 = [1, 0, 1]   # 'no'→0, 'yes'→1
print(y1)
##Approach-2
encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)
print(y_encoded)


##Result for Approach-1----------------------
model = LogisticRegression()
model.fit(X, y1)
print("Answer for approach-1:\n",model.predict([[4, 5]]),"\n")

##Result for Approach-1----------------------
model2 = LogisticRegression()
model2.fit(X, y_encoded)
print("Answer for approach-2:\n",model2.predict([[4, 5]]),"\n")
```

```python
#Q9: DecisionTreeClassifier requires numeric labels, not strings.

from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder

X_train = [[1, 2], [3, 4], [5, 6]]
y_train = ["spam", "ham", "spam"]

# Convert string labels to numeric
encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y_train)  # 'ham'->0, 'spam'->1

# Train the model
model = DecisionTreeClassifier()
model.fit(X_train, y_encoded)

# Predict example
pred = model.predict([[2, 3]])
print("Predicted (encoded):", pred)
print("Predicted (original):", encoder.inverse_transform(pred))
```
